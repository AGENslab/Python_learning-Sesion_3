# SciPy esencial — Cheatsheet

```python
import numpy as np

from scipy import stats
from scipy.optimize import curve_fit
```

## 1. Describir un conjunto de datos

```python
alturas = np.array([18.2, 19.1, 17.8, 20.3, 18.9, 19.5])

resumen = stats.describe(alturas)
print(resumen)
```

## 2. Comparar dos grupos independientes

El t-test de Welch compara las medias sin asumir varianzas iguales.

```python
control = np.array([18.2, 19.1, 17.8, 20.3, 18.9, 19.5])
tratamiento = np.array([21.0, 22.1, 20.7, 23.0, 21.8, 22.4])

resultado = stats.ttest_ind(control, tratamiento, equal_var=False)
print("t =", resultado.statistic)
print("p =", resultado.pvalue)
```

## 3. Comparar mediciones pareadas

Se usa cuando las mediciones provienen de los mismos individuos.

```python
antes = np.array([10.2, 11.1, 9.8, 10.5, 11.0, 10.8])
despues = np.array([11.4, 12.0, 10.7, 11.2, 11.9, 11.5])

resultado = stats.ttest_rel(antes, despues)
print("t =", resultado.statistic)
print("p =", resultado.pvalue)
```

## 4. Comparar tres o más grupos

ANOVA detecta si existe alguna diferencia global. Tukey identifica los pares que difieren.

```python
control = np.array([18.1, 19.0, 18.5, 19.3, 18.8, 19.1])
fert_a = np.array([20.1, 20.5, 19.8, 21.0, 20.7, 20.3])
fert_b = np.array([23.0, 22.4, 23.5, 22.9, 23.2, 22.7])

anova = stats.f_oneway(control, fert_a, fert_b)
print("ANOVA: F =", anova.statistic, "p =", anova.pvalue)

tukey = stats.tukey_hsd(control, fert_a, fert_b)
print(tukey)
```

## 5. Alternativas no paramétricas

```python
# Dos grupos independientes
mann_whitney = stats.mannwhitneyu(control, fert_a, alternative="two-sided")
print("Mann–Whitney: U =", mann_whitney.statistic, "p =", mann_whitney.pvalue)

# Tres o más grupos independientes
kruskal = stats.kruskal(control, fert_a, fert_b)
print("Kruskal–Wallis: H =", kruskal.statistic, "p =", kruskal.pvalue)
```

## 6. Ajustar una curva con `curve_fit`

```python
def logistica(t, K, r, t0):
    return K / (1 + np.exp(-r * (t - t0)))

tiempo = np.array([0, 5, 10, 15, 20, 25, 30], dtype=float)
biomasa = np.array([1.8, 3.0, 7.2, 15.0, 24.0, 30.5, 33.0])

parametros, covarianza = curve_fit(
    logistica, tiempo, biomasa, p0=[35, 0.25, 15]
)

K, r, t0 = parametros
print("K =", K)
print("r =", r)
print("t0 =", t0)
```

## Resumen

| Pregunta | Función |
|---|---|
| ¿Cómo resumo los datos? | `stats.describe` |
| ¿Difieren dos grupos independientes? | `stats.ttest_ind` |
| ¿Difieren dos mediciones relacionadas? | `stats.ttest_rel` |
| ¿Difieren tres o más grupos? | `stats.f_oneway` |
| ¿Qué pares difieren? | `stats.tukey_hsd` |
| Alternativa para dos grupos | `stats.mannwhitneyu` |
| Alternativa para tres o más grupos | `stats.kruskal` |
| ¿Cómo ajusto una función? | `curve_fit` |

> Un p-value menor que 0.05 suele considerarse evidencia contra la hipótesis nula, pero siempre debe interpretarse junto con el diseño, los supuestos y la magnitud del efecto.
