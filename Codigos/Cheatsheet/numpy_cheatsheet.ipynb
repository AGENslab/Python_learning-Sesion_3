# NumPy esencial — Cheatsheet

```python
import numpy as np
```

## 1. Crear arrays

```python
# Desde una lista
datos = np.array([12, 15, 18, 21, 24])

# Secuencias
enteros = np.arange(0, 10, 2)       # inicio, fin (no incluido), paso
puntos = np.linspace(0, 1, 5)       # 5 valores equiespaciados

print(datos)
print(enteros)
print(puntos)
```

## 2. Información, índices y cortes

```python
print("Dimensiones:", datos.ndim)
print("Forma:", datos.shape)
print("Tipo:", datos.dtype)
print("Cantidad:", datos.size)

print("Primer elemento:", datos[0])
print("Último elemento:", datos[-1])
print("Posiciones 1 a 3:", datos[1:4])
```

## 3. Operaciones vectorizadas y filtros

Las operaciones se aplican elemento por elemento sin escribir un ciclo.

```python
print(datos + 2)
print(datos * 3)
print(datos ** 2)

# Máscara booleana
mayores = datos[datos > 17]
print("Mayores que 17:", mayores)
```

## 4. Resúmenes numéricos

```python
print("Suma:", np.sum(datos))
print("Media:", np.mean(datos))
print("Mediana:", np.median(datos))
print("Mínimo:", np.min(datos))
print("Máximo:", np.max(datos))
print("Desviación estándar muestral:", np.std(datos, ddof=1))
print("Percentil 25:", np.percentile(datos, 25))
```

## 5. Arrays de dos dimensiones

```python
matriz = np.array([
    [10, 12, 14],
    [20, 22, 24]
])

print("Forma:", matriz.shape)
print("Elemento fila 0, columna 1:", matriz[0, 1])
print("Primera fila:", matriz[0, :])
print("Segunda columna:", matriz[:, 1])

# axis=0 resume cada columna; axis=1 resume cada fila
print("Media por columna:", np.mean(matriz, axis=0))
print("Media por fila:", np.mean(matriz, axis=1))
```
